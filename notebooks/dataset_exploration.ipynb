```python
import os
import cv2
import matplotlib.pyplot as plt

# Folder paths
cloud_dir = "../data/raw/RICE1/cloud"
label_dir = "../data/raw/RICE1/label"

# Get first image names
cloud_file = sorted(os.listdir(cloud_dir))[0]
label_file = sorted(os.listdir(label_dir))[0]

# Read images
cloud = cv2.imread(os.path.join(cloud_dir, cloud_file))
label = cv2.imread(os.path.join(label_dir, label_file))

# Convert BGR to RGB
cloud = cv2.cvtColor(cloud, cv2.COLOR_BGR2RGB)
label = cv2.cvtColor(label, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(cloud)
plt.title("Cloud Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(label)
plt.title("Ground Truth Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
