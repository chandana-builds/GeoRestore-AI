```python
import os
import glob
import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt

from torch.utils.data import Dataset
```

```python
class RICEDataset(Dataset):

    def __init__(self, cloud_dir, label_dir):
        self.cloud_images = sorted(glob.glob(os.path.join(cloud_dir, "*")))
        self.label_images = sorted(glob.glob(os.path.join(label_dir, "*")))

    def __len__(self):
        return len(self.cloud_images)

    def __getitem__(self, index):

        cloud = cv2.imread(self.cloud_images[index])
        label = cv2.imread(self.label_images[index])

        cloud = cv2.cvtColor(cloud, cv2.COLOR_BGR2RGB)
        label = cv2.cvtColor(label, cv2.COLOR_BGR2RGB)

        cloud = cloud.astype(np.float32) / 255.0
        label = label.astype(np.float32) / 255.0

        cloud = torch.from_numpy(cloud).permute(2, 0, 1)
        label = torch.from_numpy(label).permute(2, 0, 1)

        return cloud, label
```

```python
dataset = RICEDataset(
    "../data/raw/RICE1/cloud",
    "../data/raw/RICE1/label"
)

print("Total Images:", len(dataset))

cloud, label = dataset[0]

print("Cloud Tensor Shape:", cloud.shape)
print("Label Tensor Shape:", label.shape)
```

```python
from torch.utils.data import DataLoader
```

```python
train_loader = DataLoader(
    dataset,
    batch_size=8,
    shuffle=True,
    num_workers=0
)
```

```python
cloud_batch, label_batch = next(iter(train_loader))

print("Cloud Batch Shape :", cloud_batch.shape)
print("Label Batch Shape :", label_batch.shape)
```
