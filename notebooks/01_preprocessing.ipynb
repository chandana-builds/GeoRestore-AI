```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Dataset paths
cloud_dir = "../data/raw/RICE1/cloud"
label_dir = "../data/raw/RICE1/label"

# Read the first image
cloud_file = sorted(os.listdir(cloud_dir))[0]
label_file = sorted(os.listdir(label_dir))[0]

cloud_path = os.path.join(cloud_dir, cloud_file)
label_path = os.path.join(label_dir, label_file)

cloud = cv2.imread(cloud_path)
label = cv2.imread(label_path)

# Convert BGR to RGB
cloud = cv2.cvtColor(cloud, cv2.COLOR_BGR2RGB)
label = cv2.cvtColor(label, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(cloud)
plt.title("Cloud Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(label)
plt.title("Ground Truth")
plt.axis("off")

plt.show()
```

```python
print("Cloud Shape :", cloud.shape)
print("Label Shape :", label.shape)

print("Cloud Data Type :", cloud.dtype)
print("Label Data Type :", label.dtype)

print("Cloud Min Pixel :", cloud.min())
print("Cloud Max Pixel :", cloud.max())
```

```python
# Normalize images to the range [0, 1]
cloud_normalized = cloud.astype(np.float32) / 255.0
label_normalized = label.astype(np.float32) / 255.0
```

```python
print("Cloud Image")
print("Shape:", cloud_normalized.shape)
print("Data Type:", cloud_normalized.dtype)
print("Min:", cloud_normalized.min())
print("Max:", cloud_normalized.max())

print("\nGround Truth")
print("Shape:", label_normalized.shape)
print("Data Type:", label_normalized.dtype)
print("Min:", label_normalized.min())
print("Max:", label_normalized.max())
```
